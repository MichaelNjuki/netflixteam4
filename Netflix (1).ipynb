```python
class Movie:
    def __init__(self, movie_id, title, genre):
        self.movie_id = movie_id
        self.title = title
        self.genre = genre
```

```python
class User:
    def __init__(self, user_id, name):
        self.user_id = user_id
        self.name = name
```

```python
class WatchHistory:
    def __init__(self, user_id, movie_id, rating):
        self.user_id = user_id
        self.movie_id = movie_id
        self.rating = rating
```

```python
movies = [
    Movie(1, "The Matrix", "Sci-Fi"),
    Movie(2, "Inception", "Sci-Fi"),
    Movie(3, "Interstellar", "Sci-Fi")
]
```

```python
users = [
    User(101, "Ada"),
    User(102, "Charles"),
    User(103, "Linus")
]
```

```python
history = [
    WatchHistory(101, 1, 5),
    WatchHistory(101, 2, 4),
    WatchHistory(102, 1, 5),
    WatchHistory(103, 3, 5),
    WatchHistory(103, 1, 4),
    WatchHistory(103, 2, 3)
]
```

```python
total_views = len(history)

print("Total views:", total_views)
```

```python
from collections import Counter

movie_views = Counter()

for h in history:
    movie_views[h.movie_id] += 1

print(movie_views)
```

```python
active_users = set()

for h in history:
    active_users.add(h.user_id)

print("Active users:", len(active_users))
```

```python
from collections import defaultdict

ratings = defaultdict(list)

for h in history:
    ratings[h.movie_id].append(h.rating)
```

```python
for movie_id, values in ratings.items():
    average = sum(values) / len(values)

    print(movie_id, round(average, 2))
```

```python
movie_names = {}

for m in movies:
    movie_names[m.movie_id] = m.title

for movie_id, count in movie_views.most_common():
    print(movie_names[movie_id], count)
```

```python
print("\n= NETFLIX DAILY REPORT =")

print("Total viewing activity:", total_views)

print("Active users:", len(active_users))

print("\nMost watched movies:")

for movie_id, count in movie_views.most_common():
    print(movie_names[movie_id], "-", count, "views")

print("\nAverage ratings:")

for movie_id, values in ratings.items():
    average = sum(values) / len(values)
    print(movie_names[movie_id], "-", round(average, 2))

print("\n = REPORT COMPLETED =")
```
