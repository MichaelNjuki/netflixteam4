```python
class Movie:
    def __init__(self, movie_id, title, genre):
        self.movie_id = movie_id
        self.title = title
        self.genre = genre

class User:
    def __init__(self, user_id, name):
        self.user_id = user_id
        self.name = name

class WatchHistory:
    def __init__(self, user_id, movie_id, rating):
        self.user_id = user_id
        self.movie_id = movie_id
        self.rating = rating
```

```python
def transform(movies, users, history):

    # 1. Total number of watches
    total_events = len(history)

    # 2. Average rating
    total_rating = 0
    for watch in history:
        total_rating = total_rating + watch.rating
    average_rating = total_rating / total_events

    # 3. Count how many times each movie was watched
    movie_count = {}
    for watch in history:
        movie_id = watch.movie_id
        if movie_id in movie_count:
            movie_count[movie_id] = movie_count[movie_id] + 1
        else:
            movie_count[movie_id] = 1

    most_watched_id = max(movie_count, key=movie_count.get)
    most_watched_count = movie_count[most_watched_id]

    most_watched_title = ""
    for movie in movies:
        if movie.movie_id == most_watched_id:
            most_watched_title = movie.title

    # 4. Count how many times each user watched something
    user_count = {}
    for watch in history:
        user_id = watch.user_id
        if user_id in user_count:
            user_count[user_id] = user_count[user_id] + 1
        else:
            user_count[user_id] = 1

    most_active_id = max(user_count, key=user_count.get)
    most_active_count = user_count[most_active_id]

    most_active_name = ""
    for user in users:
        if user.user_id == most_active_id:
            most_active_name = user.name


    print("NETFLIX  VIEWING REPORT")
    print("Total Viewing Events :", total_events)
    print("Average Rating       :", round(average_rating, 2))
    print("Most Watched Movie   :", most_watched_title, "(", most_watched_count, "views)")
    print("Most Active User     :", most_active_name, "(", most_active_count, "watches)")
    
```

```python
transform(movies, users, history)
```
